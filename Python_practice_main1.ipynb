# Simple Calculator

```python
class Calculator:
    def __init__(self):
        pass
    
    def add(self, num1, num2):
        return num1+num2
    
    def subtraction(self, num1, num2):
        return num1-num2
        
    def multiplication(self, num1, num2):
        return num1*num2
        
    def division(self, num1, num2):
        try:
            return num1/num2
        except Exception as e:
            print(e)
#             raise ZeroDivisionError("Denominator Must not be a '0' ")
    
    def inp(self):
        try:
            num1 =  float(input("Enter first number : "))
            num2 =  float(input("Enter second number : "))
            return num1, num2
        except Exception as e:
            print(e)
#             rasie ValueError("Invalid input, Please Enter Numbers")
    def On(self):
        while True:
            try:
                print("Select operation:")
                print("1. Add")
                print("2. Subtract")
                print("3. Multiply")
                print("4. Divide")
                print("5. Exit")
                
                ch =  input("Enter choice (1/2/3/4/5): ")
                if ch=="5":
                    print("Exiting ")
                    break
                
                num1, num2 = self.inp()
                
                if ch=="1":
                    print(f"Addition of {num1} and {num2} is : {self.add(num1, num2)}")
                elif ch=="2":
                    print(f"Subtraction of {num1} and {num2} is : {self.subtraction(num1, num2)}")
                elif ch=="3":
                    print(f"Multiplication of {num1} and {num2} is : {self.multiplication(num1, num2)}")
                elif ch=="4":
                    print(f"Division of {num1} and {num2} is : {self.division(num1, num2)}")
                print("==="*15)
                
            except ValueError as e:
                print(e)
            except Exception as e:
                print(f"An unexpected error occurred: {e}")
                
                    
task1 =  Calculator()
task1.On()
```

# Number Guessing Game

```python
import random
class Guessing_game:
    def __init__(self, low=1, high=50):
        self.low = low
        self.high = high
        self.num = random.randint(self.low, self.high)
        
    def start(self):
        print("=="*5)
        print("Game Starts")
        print(f"Guess the number in the range {self.low} and {self.high}")
        
        count = 0
        while True:
            try:
                guess =  int(input("Enter a number: "))
                count+=1
                
                if guess < self.low or guess > self.high:
                    print(f"Please Enter between {self.low} and {self.high}")
                elif guess < self.num:
                    print("Too low, Try higher")
                elif guess > self.num:
                    print("Too high, Try lower")
                else:
                    print(f"You guess the number '{guess}' correct in '{count}' guesses")
                    break
            except ValueError:
                print("Invalid input! Please enter a valid number.")
                
task2 =  Guessing_game()
task2.start()
```

```python
# Count the lines, words, characters in a file
para = """File handling in programming 
refers to the process of creating, reading, writing, and manipulating files on a computer's file system. 
In Python, this is typically done using built-in functions and methods that allow developers to interact with files seamlessly.
Files can be opened in various modes, 
such as read ('r'), write ('w'), append ('a'), or binary mode ('b'). Once a file is opened, 
data can be read or written using methods like read(), readline(), write(), or writelines(). 
Proper file handling also involves closing files after operations 
using the close() method to free up system resources and ensure data integrity. 
Additionally, using the with statement in Python provides a convenient way to handle files, 
as it automatically takes care of closing the file, even if an error occurs, 
promoting better resource management and cleaner code. 
Overall, effective file handling is crucial for data persistence and manipulation in many applications.
"""
with open("./demo1.txt", "w") as f:
    f.write(para)
    
with open("./demo1.txt", "r")as f2:
    print(f2.read())
    
    
class Countval:
    def __init__(self, path):
        self.path = path
    def count_values(self):
        try:
            with open(self.path, "r")as f:
                line_count = 0
                word_count = 0
                char_count = 0
                for line in f:
                    line_count+=1
                    char_count += len(line)
                    word_count += len(line.split())
                return line_count, word_count, char_count
        except FileNotFoundError:
            print(f"The file '{self.path}' does not exist.")
            return
filePath = "./demo1.txt"
task3 = Countval(filePath)
lines, words, chars = task3.count_values()
print(f"Lines in {filePath} are '{lines}'")
print(f"Words in {filePath} are '{words}'")
print(f"Chars in {filePath} are '{chars}'")
```
