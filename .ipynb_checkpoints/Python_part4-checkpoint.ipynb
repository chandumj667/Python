# Object Oriented Programming

#### Classes and Objects

```python
# Class     : It defines the attributes(variables) and methods(functions) for the object 
# Object    : It is an instance of class. It holds both data (attributes) and behaviour(methods) 
# the memory of the class (structure) is allocated when the object creation only 

# class
class sec1:
#     it can have the attributes and methods in it
    x = 10


# Object creation and accessing the values from the class
valu = sec1.x
print(valu)
```

```python
# __init__()
# all classes have a function called __init__(). which is always being executed when the class is initiated
# in other prog langs it is like default constructor
# it sould have a parameter called "self" or you can name it as your wish 
# self : It allows us to distinguish between instance attributes and local variables.
# Using self, you can assign instance-specific attributes and access them across other methods of the class.

class Person:
      def __init__(self, fname, lname, organization, age):
        self.fname = fname
        self.lname = lname
        self.organization = organization
        self.age = age

p1 = Person("Tony", "Stark", "Stark Industries", 36)

print(f"First Name\t: {p1.fname}")
print(f"Last Name\t: {p1.lname}")
print(f"Organization\t: {p1.organization}")
print(f"And his age\t: {p1.age}")
```

```python
# __str__() : it controls what should be return if the class object is represented as string

class Person:
    def __init__(self, fname, lname, organization, age):
        self.fname = fname
        self.lname = lname
        self.organization = organization
        self.age = age
    def __str__(self):
        return f"{self.fname} {self.lname} working in '{self.organization}' and his age is {self.age}"

p1 = Person("Tony", "Stark", "Stark Industries", 36)
print(p1)
```

```python
# methods inside the class and it can be accessed by the object
class Student:
    def __init__(self, name, ID, mail):
        self.name = name
        self.ID = ID
        self.mail = mail
    def score(self, maths, physics, chemistry):
        self.maths = maths
        self.physics = physics
        self.chemistry = chemistry
        self.percentage = (self.maths+self.physics+self.chemistry)/3
        if self.percentage>70:
            print("Grade : A")
        elif 50 <self.percentage<=35:
            print("Grade : B")
        else:
            print("Grade : C")
    def __str__(self):
        return f"{self.name} with ID '{self.ID}' scored {self.percentage:.2f}%"
    
std1 = Student("Tony", "12342", "tony@gmail.com")
std1.score(92, 94, 98)
print(std1)
```

```python
# modify the object properties 
std1.ID = 202021
print(std1)
```

```python
# del object properties
# del std1.mail
# print(std1)

if std1.mail:
    print(f"Student mail : {std1.mail}")
else:
    print("No mail ID")

# del object property
del std1.mail
# del object
del std1

print(std1)
```

# Encapsulation

```python
# Encapsulation refers to bundling data (attributes) and methods (functions) 
# that operate on the data into a single unit or class. 
# It restricts access to some of the object's components, which is known as data hiding.

# In Python, encapsulation can be achieved using private and public attributes and methods.

# A single underscore _ suggests that the attribute is intended to be private.
# A double underscore __ makes the attribute private to the class and not accessible directly.


class Student:
    def __init__(self, name, age):
        self.name = name  
        self.__age = age  # Private attribute

    
    def get_info(self):
        return f"Name: {self.name}, Age: {self.__age}"

    # Private method
    def __private_method(self):
        return f"This is private and Name : {self.name}, Age: {self.__age}"

std1 = Student("rogerrs", 30)

print(std1.get_info())  # Accessing public method
print(std1.__private_method())  # Error: Trying to access private method
```

```python
# multiple inhertance :
# inheriting from different parent classes
class Parent1:
    def method1(self):
        print("Parent1 method")

class Parent2:
    def method2(self):
        print("Parent2 method")

class Child(Parent1, Parent2):
    def method3(self):
        print("Child method")
        
child1 = Child()
child1.method1()
child1.method2()
child1.method3()
```

```python
# multi-level inhertance
# deriving child class from parent which is derived from its parent
# grand-parent --> parent --> child
class Grandparent:
    def method1(self):
        print("Grandparent method")

class Parent(Grandparent):
    def method2(self):
        print("Parent method")

class Child(Parent):
    def method3(self):
        print("Child method")

child2 = Child()
child2.method1()
child2.method2()
child2.method3()
```

```python
# Hierarchial inheritance
#  from one parent two or more classes derived
class Parent:
    def method1(self):
        print("Parent method")

class Child1(Parent):
    def method2(self):
        print("Child1 method")

class Child2(Parent):
    def method3(self):
        print("Child2 method")

        
child3 = Child2()
child3.method3()
child3.method1()

child4 = Child1()
child4.method2()
child4.method1()
```

# Polymorphism

```python
# Polymorphism - means many forms, 
# in programming the  methods, function, operators having same name that can executed on many objects and classes
# ex1: "+" : in numbers it is used to sum them but in characters or strings it is used to concatinate them
# ex2: "len()" : in string it return the num of chars, but in tuple or list or dict it return the number of elements in it

class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def move(self):
        print("Move!")

class Car(Vehicle):
    pass

class Boat(Vehicle):
    def move(self):
        print("Sail!")

class Plane(Vehicle):
    def move(self):
        print("Fly!")

# here move function is move() function is overrided for each object 
car1 = Car("Ford", "Mustang") #Create a Car object
boat1 = Boat("Ibiza", "Touring 20") #Create a Boat object
plane1 = Plane("Boeing", "747") #Create a Plane object

for x in (car1, boat1, plane1):
    print(x.brand)
    print(x.model)
    x.move()
    print()
```

# Abstraction

```python
# Abstraction means hiding the complex implementation details and showing only the necessary parts. 
# Python achieves abstraction using abstract classes. 
# Abstract classes are classes that cannot be instantiated and contain abstract methods, 
# which must be implemented by subclasses.


# from abc import ABC, abstractmethod

# Abstract Class
class Vehicle(ABC):
#     @abstractmethod
    def start_engine(self):
        pass

class Car(Vehicle):
    def start_engine(self):
        return "Car engine started"

class Bike(Vehicle):
    def start_engine(self):
        return "Bike engine started"

car = Car()
bike = Bike()

print(car.start_engine())
print(bike.start_engine())
```

# Method Overloading

```python
# When a child class provides its own implementation of a method that is already defined in its parent class, 
# it's called method overriding. 
# It allows the child class to change the behavior of a method inherited from the parent class.

class Parent:
    def show_message(self):
        return "This is a message from Parent"

class Child(Parent):
    # Overriding the parent method
    def show_message(self):
        return "This is a message from Child"

child = Child()
print(child.show_message())  
```

# Method Overriding

```python
# Python doesn't support method overloading (same method name with different arguments) in a traditional sense. 
# However, we can achieve a similar effect using default arguments or variable-length arguments (*args).

class Calculator:
    def add(self, a, b, c=0):  
        return a + b + c

calc = Calculator()
print(calc.add(2, 3))       
print(calc.add(2, 3, 4))
```
