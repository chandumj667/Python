```python
#9 Zero Converter
"""
You are given a number n. The number n can be negative or positive. 
If n is negative, print numbers from n to 0 by adding 1 to n in the neg function. 
If positive, print numbers from n-1 to 0 by subtracting 1 from n in the pos function.

Example 1:
Input:
n = -3
Output:
-3 -2 -1 0

Example 2:
Input:
n = 4
Output:
3 2 1 0
"""

def print_num(num):
    if num<0:
        while(num<=0):
            print(num, end=" ")
            num+=1
    elif num>0:
        num-=1
        while(num>=0):
            print(num, end=" ")
            num-=1
    else:
        print(num)
        
print_num(6)
print()
print_num(-6)
print()
print_num(0)
```

```python
#10 Check Prime Number
"""
Given a positive integer N, the task is to write a Python program to check if the number is Prime or not
"""

#  we can loop till sqrt(n) 
import math
num = int(input("Enter a number: "))

def prime_or_not(num):
    if(num>1):
        for i in range(2, int(math.sqrt(num))+1):
            if num%i ==0:
                return False
        return True
    else:
        return False

print(f"{num} is {'Not a prime' if not prime_or_not(num) else 'Prime'}")
```

```python
# 19 reverse words
"""
Given a String str, reverse the string without reversing its individual words. Words are separated by dots.

Note: The last character has not been '.'. 

Examples :
Input: str = i.like.this.program.very.much
Output: much.very.program.this.like.i
"""
def rev(lst):
    str3 =""
    for i in range(len(lst)-1, -1, -1):
        str3+=''.join(lst[i])+'.'
    return str3



def reverse_words(lst: list)->str:
    str2 = ""
    for i in lst:
        str2 +=''.join(i[::-1])+'.'
    return str2
        

str1 = "i.like.this.program.very.much"
lst = str1.split(".")
print(str1)

print(rev(lst))

print(reverse_words(lst))
```

```python
# 22 Repeat the Strings
"""
Given two strings a and b. 
The task is to make a new string where the string with longer length 
should be in between and the one with shorter length should be outside on front and end. 
New string should be like shorter+longer+shorter.

Example:
Input: 
a = Hi, b = There
Output: 
HiThereHi
"""

def repeat_str(a, b)->str:
    short=""
    longer=""
    if len(a)<len(b):
        short += a
        longer += b
    else:
        short += b
        longer += a
    return short+longer+short

a = "Hi"
b = "There"
print(repeat_str(a, b))
```

```python
# 23 Palindrome String
"""
Given a string S, check if it is palindrome or not.

Example 1:
Input: 
S = "abba"
Output: 
abba is a palindrome
"""

def check_palindrome(a)->bool:
#     if a==a[::-1]:
#         return True
#     else:
#         return False
    l = 0
    h = len(a)-1
    while l<h:
        if a[l]!=a[h]:
            return False
        else:
            return True
        l+=1        
        h-=1

a ="Hellolleh"
if check_palindrome(a.lower()):
    print(f"'{a}' : is a palindrome")
else:
    print(f"'{a}' : is not a palindrome")
```

```python
# 24 Remove all duplicates from a given string
"""
Given a string str which may contain lowercase and uppercase characters. 
The task is to remove all duplicate characters from the string and find the resultant string. 
The order of remaining characters in the output should be same as in the original string.

Example 1:
Input:
str = "geEksforGEeks"
Output: 
"geEksforG"
"""


def remove_dups(strr)->str:
    s=""
    for i in strr:
        if i not in s:
            s+=i
    return s
strr = "geEksforGEeks"
print(f"{strr} : {remove_dups(strr)}")
```

```python
# 25 Check Anagram String
"""
Given two strings s1 and s2 consisting of lowercase characters. 
The task is to check whether two given strings are an anagram of each other or not. 
An anagram of a string is another string that contains the same characters, 
only the order of characters can be different. For example, act and tac are an anagram of each other. 
Strings s1 and s2 can only contain lowercase alphabets.

Note: You can assume both the strings s1 & s2 are non-empty.

Examples :

Input: s1 = "listen", s2 = "silent"
Output: true
"""


def check_anagrams(s1:str, s2:str)->bool:
    if len(s1)!=len(s2):
        return False
    if sorted(s1)!=sorted(s2):
        return False
    else:
        return True
    
s1 = "listen"
s2 = "silent"
if check_anagrams(s1, s2):
    print(f"'{s1}' and '{s2}' are Anagrams")
else:
    print(f"'{s1}' and '{s2}' are Not Anagrams")
    
```

```python
# 27 Maximum and Minimum K elements in Tuple
"""
Given a tuple with numerical values in it  and a value 'k'. no duplicates
The task is to return k minimum and k maximum elements from the tuple
Input : test_tup = (3, 7, 1, 18, 9), k = 2 
Output : (3, 1, 9, 18)
"""

def min_max_k(tup:tuple, k:int)->tuple:
    if k>len(tup):
        return tup
    
    # if there are no duplicate values means we can sort the tuple and return 0, 1 and n-1, n-2 index values
    srtd_tup = tuple(sorted(tup))
    print(f"{k} - minimum values: {srtd_tup[:k]}")  
    print(f"{k} - maximum values: {srtd_tup[-k:]}") 

    # If there are duplicate values, we need to convert the tuple to a set first
    set_tup = set(tup)  
    set_tup = tuple(sorted(set_tup))  # Sort and convert back to tuple
    print(f"{k} - minimum values (if duplicates): {set_tup[:k]}")  # First k minimum values
    print(f"{k} - maximum values (if duplicates): {set_tup[-k:]}") 
    
test_tup1 = (3, 7, 1, 18, 9)
test_tup2 = (3, 3,17, 23, 23, 7, 1, 18, 9)

k=2
min_max_k(test_tup1, k)
min_max_k(test_tup2, k)

    
```

```python
# 34 set operations
"""
Some important functions in Set in Python:
add(): Adds an element to the set.
clear(): Removes all elements from the set
discard(): Removes an element from the set if present.
remove(): Removes an element from the set. If the element is not present, it raises error.
pop(): Removes and returns an arbitary set element. Raise error if the set is empty.
union(): Returns the union of sets in a new set
update(): Updates the set with the union of itself and others
len(): Return the length of set.
sorted(): Return a new sorted list from elements in the set.
sum(): Return the sum of all elements in the set.

Let's implement these methods through a question. Given Q queries to do some operation on Set, 
the task is to perform all the queries in the Set as given below:
a. i element: Adds element to the set.
b. r element: Remove element from set.
c. s: Find sum of elements in set. Returns the sum of elements in Set.
"""


def insert(s, element):
    s.add(element)
    return s
    
def remove_from_set(s, element):
    s.discard(element)
    return s
    
def sum_set(s):
    return sum(s)


s = {1, 2, 3, 4, 5, 6, 7, 8, 9}
ele =int(input("Enter an element"))
print(f"Inserting element {ele} : {insert(s, ele)}")
print(f"Removing element {ele}: {remove_from_set(s, ele)}")
print(f"Summation {s} : {sum_set(s)}")
                                   
```

```python
# 37 Password validation 
"""
You will be provided a string str which you have to validate.
The validation rules are as follows:

The string is valid only if it starts with lowercase characters, 
followed by special characters !@#$% followed by numbers.
In any other case the string is not valid.
Example:
Input: 
str = asdsab@!@234
Output: 
True
"""


def password_valid(str):
    pat = "[a-z]+[!@#$%]+[0-9]"
    match = re.search(pat, str)
    if match:
        return True
    else:
        return False
str1 = "asdsab@!@234"  
str2 = "steve36"  

print(f"{'Valid Password' if password_valid(str1) else 'Invalid Passowrd'}")
print(f"{'Valid Password' if password_valid(str2) else 'Invalid Passowrd'}")
```
