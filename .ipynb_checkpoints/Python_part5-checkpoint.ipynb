# DECORATORS

```python
# A decorator is a function that takes another function as an argument, 
# adds some functionality to it, and returns a new function (or the same function) with the added functionality.

def decorator1(func):
    def inner(val1, val2):
        result = func(val1, val2)
        return f"Using decorators: {result}"
    return inner


@decorator1
def multiply(x, y):
    return x*y

print(multiply(10, 20))
```

```python
def decorator2(func):
    def wrapper():
        print("Decorator 2")
        func()
    return wrapper

def decorator3(func):
    def wrapper():
        print("Decorator 3")
        func()
    return wrapper

@decorator3
@decorator2
def say_hi():
    print("Hi!")

say_hi()
```

# JSON

```python
# JSON is a syntax for storing and exchanging data.
# JSON is text, written with JavaScript object notation.

# Parse JSON - Convert from JSON to Python

import json

x =  '{ "name":"John", "age":30, "city":"New York"}'
y = json.loads(x)

print(y["age"])


# Convert from Python to JSON
z = json.dumps(x)
print(z)
print()

print(json.dumps({"name": "John", "age": 30}))
print(json.dumps(["apple", "bananas"]))
print(json.dumps(("apple", "bananas")))
print(json.dumps("hello"))
print(json.dumps(42))
print(json.dumps(31.76))
print(json.dumps(True))
print(json.dumps(False))
print(json.dumps(None))
```

# RegEX

```python
# Regex :  Regular Expression
# It is a sequence of characters that forms a search pattern
# It is used to check if a string contains the specified search pattern or not

import re # to use regex need to import re module

# Special Sequences
# \A : Returns a match if the specified chars are at the beginning of the string
# \Z : Returns a match if the specified chars are at the end of the string
# \b : Returns a match if the specified chars are at the beginning or end of the string
# \B : Returns a match if the specified chars are present but not at the beginning or end of the string
# \d : Returns a match if the string contains digits [0-9]
# \D : Returns a match if the string does not contains digits [0-9]
# \s : Returns a match if the string contains whitespace chars
# \S : Returns a match if the string does not contains whitespace chars
# \w : Returns a match if the string contains word chars (chars from a to z, digits from 0 to 9)
# \W : Returns a match if the string does not contains word chars (chars from a to z, digits from 0 to 9)


# Metacharacters
# []  : A set of characters "[a-m]"
# .   : Any character (except newline character) "he..o"
# ^   : Starts with specified char or word  "^hello"
# $   : Ends with specified char or word "planet$"
# *   : Zero or more occurrences "he.*o"
# +   : One or more occurrences "he.+o"
# ?   : Zero or one occurrences "he.?o" 
# {}  : Exactly the specified number of occurrences	"he.{2}o"
# |   : Either or "x|y"
# ()  : Capture and group

# functions()

# findall() : it retunrs a list of all occurances of specified char or string from a string
# if no match was found it returns empty string
sentence = "Hi all this is Captain America where he also served as Captain in WWII and Captain to a Aircraft"
sentence2 = "Hi all this is Captain America where he also served as Captain in WWII and Captain to a Aircraft"
ans1  = re.findall("Captain", sentence)
print(f"Using findall() : {ans1}")


# search() : function searches the string for a match, and returns a Match if there is a match 
# if there are more than one accourance it returns first occurance
ans2 = re.search("Captain", sentence)
# as it returns the match object so returning the index of the match
print(f"Using search() : {ans2.start()}")

# split() : return a list of values where the entire string got split by specified char
ans3 = re.split("\s", sentence)
print(f"Using split() : {ans3}")

# we can specify the split as using maxsplit parameter
# it make the sentence x+1 splits where x is given paramenter and the remaining sentence takes as 1 split
ans4 = re.split("\s", sentence, 4)
print(f"Using split() and maxsplit: {ans4}")


# sub() : it replaces the match with specified word or char and we can limit how many changes to do
ans5 = re.sub("\s", "_", sentence)
print(f"Using sub() : {ans5}")

ans6 = re.sub("\s", " || ", sentence, 5)
print(f"Using sub() with count : {ans6}")
```

# Exception Handling

```python
# Errors : 
# 1) Compile time error : when there are any syntax errors it leads to compile time error
# 2) Run time error : syntax is correct but, it throughs an error at run time and stops the next lines of code
# to handle these run time errors we have keywords like try, except, else, finally and raise.
# In try block we write the critical where may be the errors happen 
# syntax:
a = 10
b = 0
# print(a/b)
try:
    print(a/b)
except Exception as e:
    print(e)
    
# the next line didnt stop running as the code is handled by exception handlers
print(a+b)
```

```python
#  we can use except n number of times to handle diff types of errors
# try: it takes care about the critical code
# except: if any errors happens except case statements handles and continue to the next lines of code
# else: if no errors were raised means no except statements are encountered then else statement is executed
# finally: whether the code execution happens in try, except or not but finally case statements definetly execute
# finally most used in file handling or db connections to close the db connection if any error happends in critical code
# rasie : 
x = "Captain America"
try:
  print(x)
except NameError:
  print("Variable x is not defined")
except ValueError:
  print("Something else went wrong")
except IndexError:
    print("Not in range")
else:
    print(f"X : {x}")
finally:
    print("at last this statement definetly executed")
    
print()
try:
    print(x+2)
except ValueError:
    print("x is not defined")
except TypeError:
    print("x is not a number")
else:
    print(X*3)
finally:
    print(x)
    
print()
if(len(x)<16):
    raise Exception ("Name should be greate than 15 chars ")
else:
    print(x)
    
type(x)
print(len(x))
```

```python
try:
    num = int(input("Enter a number: "))
    result = 100 / num
except ValueError:
    print("Invalid input! Please enter an integer.")
except ZeroDivisionError:
    print("Division by zero is not allowed.")
else:
    print(f"The result is {result}")
finally:
    print("This will always execute, regardless of exceptions.")
```

# File Handling

```python
# file handling is the main concept in python to handle the file operations (read, write) and to access the different files.
# open() : to access any file
# "r" : default file operation, It reads a file when open, if the file doesnot exits throughs an error
# "w" : It writes the content in to a specified file when open, 
#     if the file doesnot exits it create a new file, if already exists it override the conten in the existing file
# "a" : appends the text to a specified file, if file exits its add the new data, if file doesnot exists it create one 
# "x" : it creates a specified file, if the file already exists it returns an error.
# "t" : default mode : text mode: opens the file in text mode.
# "b" : binary mode : open the file in binary format ("rb": "read binary", "wb": "write binary", "ab": "append binary")

# syntax:
file1 = open("./timecomplexity.txt", "r")
print(file1.read())

print("-"*100)
with open("./timecomplexity.txt", "r") as file2:
    print(file2.read())
    print("***"*10)
    print(file2.readline())
#     .read(number) : it reads the "number" of characters starting from the cursor points 
```

```python
# .readline() :  reads each line in the file
# .readlines() : reads all the lines and returns in a list of lines
with open("./timecomplexity.txt", "r") as file3:
    print(file3.readline())
    print(file3.readline())
    print(file3.readline())
    
```

```python
# count number of lines in a file 
with open("./timecomplexity.txt", "r") as f:
    count_line =0
    count_words =0
    count_chars =0
    
    for line in f:
        count_line +=1
        words = line.split()
        count_words += len(words)
        count_chars += len(line)
    print(f"Lines in file {f} \t: {count_line}")
    print(f"Words in file {f} \t: {count_words}")    
    print(f"Chars in file {f} \t: {count_chars}")    
```
