# Lists [] and Operations 

List is a datastructure that stores elements/values in an order.
we can easily access the elements using their index values.
List are mutable, we can store different data types in it.

```python
lst1 = ['apple', 'banana', 'grapes', 'strawberry', 'avacado']
print(lst1)
lst2 = [123, 34.23, True, 'apple']
print(lst2)
```

```python
# len() : to find the length of the list
print(len(lst1))
print(len(lst2))

# list constructor : we can conver or declare the anytype of data into a list
new_lst = list(("1", "2", "3"))
print(f"new_lst: {new_lst} and it is a type of {type(new_lst)}")


# accessing list elements
print(lst1[1]) #the indexing is starts from 0
print(lst1[-1]) # using negative indexing we can access the elements from the end of the list
# print(lst1[start_index:end_index]) in between the start_index and end_index we can access the elements
print(lst1[1:5])
print(lst1[:4])
print(lst1[-4:-1])
print(lst1[-4:])

```

### changing the elements in the list 

```python
names =["Sai", "Chandu", "Krishna", "Jagadeesh", "Swamy", "Narsi", "Sandy"]
print(names)
names[1] = "Chandra Sai" #by using index we can change the element or value in the index
print(names)

names[1:4] = ["Hari", "Ram", "Abhi"]  #if we give more eles than the range the new ele replaces the oldones in that place
print(names)


# insert(index, value_to_be_insert)
names.insert(len(names), "Subbhu")
print(names)
names.insert(2, "Sahith")
print(names)

# lst.append(value):  it appends the values at the end of the list
names.append(["Gopi", "Balu"])
print(names)

# + or .extend() : to concatinate two lists 
lst3 = lst1+lst2
print(lst3)

lst2.extend(lst1)
print(lst2)

# .remove() : to remove the first occurance of an item in the list
lst2.remove("avacado")
print(lst2)

# .pop(index) : to remove an item from a specified index
print("\n",names)
names.pop(2)
print(names)

# to remove an item from an index as like pop() we can also use "del lst[index]"
# and also we can "del lst" to del the entire list
# del lst2
#  to clear the contents in the list we use lst.clear()
print("\n", lst3)
lst3.clear()
print(lst3)
```

### looping through the list

```python
# using "in"
for item in lst1:
    print(item)

print()
# using len and range : len gives the length of the list and range makes iteration from starting to end 
for i in range(len(lst1)):
    print(i, lst1[i])
    
# using list comprehension
print()
new_lst1 = [print(x) for x in lst1]
```

### List comprehension
list comprehension is the way to create a new list from the elements in the existing list 

```python
fruit_basket = ["Apple", "Banana", "Mango", "Orange", "Strawberry", "Pineapple", "Blueberry", "Watermelon", "Grapes", "Peach", "Kiwi", "Pomegranate", "Cherry", "Papaya", "Raspberry"]
word_len = [(len(fruit), fruit) for fruit in fruit_basket if len(fruit)>5]
print(len(word_len))
[print(item) for item in word_len]
not_apple = [item for item in fruit_basket if item != "Apple"]
print()
fruit_upper = [print(item.upper()) for item in fruit_basket]

not_apple
```

### sort lists

```python
# Ascending order
nums1 = [10, 20, 30, 40, 50, 60]
names = ['chandu', 'arjun', 'krishna', 'Bheema']
nums1.sort()
print(nums1)
names.sort()
print(names)


# Descending order
nums1.sort(reverse=True)
print(nums1)
names.sort(reverse=True)
print(names)

names.sort(key = str.lower)
print(names)
```

### copy list

```python
listA = ["ABC", "def", 123, 45.67, True]
listB = listA
print(listB)

listC = listA.copy()
listC.append("Chocolate")
print(listC)

listD = list(listC[1:-2])
print(listD)
```

### Joining lists

```python
lst_a = listA+listB
print(lst_a)
lst_b=[]
listD.extend(listC)
lst_b.extend(listD)
print(lst_b)

# count() : return the number of specified elements in the list
print(lst_b.count("ABC"))
```

# TUPLE ()

```python
# Tuple is a datastructure in python
# it is ordered and unchangable (immutable)
# once we assigned a value or element in to a tuple we can not change or add or remove 
# elements can be accessed from the 
```

```python
tup = ("abc", "DEF", 123, 43.2, True)
print(tup)
print(len(tup))

# accessing tuple elements by index values
print(tup[0])
print(tup[-2])

# range of index as like list
print(tup[1:5])
print(tup[-5:-1])
```

```python
#  to update the values in tuple we cannot directly do that
# first convert the tuple in to list and update the list with the value
# then convert back the list to tuple
#  for any changeble operation we need to convert the tuple into list and then back to tuple

lst_tup = list(tup)
lst_tup.extend(["appple", "banana", "grapes", "orange"])

tup = tuple(lst_tup)
print(tup)
```

```python
#  tuple unpack : assigning tuple values to different varibles
tup1 = ("apple", "oranage", "banana")
fruit1, fruit2, fruit3 = tup1
print(fruit1, " ", fruit2, " ", fruit3)

val1, val2, *val3 =  tup
print(val1, "| ", val2, "| ", *val3)
```

# Sets {}

```python
# set is a immutable, unordered, unindexed collection
# set doesn't store duplicate items in it
# in set both True and 1, False and 0 are treated as same
```

```python
# a set can contain different data types
st1 = {"apple", "orange", "greapes", 1, 231, 3.43, True}
print(st1)

# len(): used to find the length of the set
print(len(st1))

# set() constructor

st2 = set(("chandu", "krishna", "Abhi", "nandu", "jithu", 250, 78.12))
print(st2)
```

### Access set items  

```python
#  we cannot access the set elements directly as they dont have any index 
for val in st2:
    print(val)
    
print()
# using in and not in
print("apple" in st1)
print("kiwi" in st1)
```

### adding new elements to the existing set 

```python
#  add() : adding one element 
st1.add("Kiwi")
print(st1)

# .update() add list items to a set
lst = ["banana", "mango", "berries"]
st1.update(lst)

print(st1)
```

### remove elements  

```python
# .remove(), .discard(), .pop(), .clear()
# remove() : if the  item to be removed not in the set then remove() will raise an error
# discard() : if the item to be removed not in the set then discard() does not raise an error
# pop() :  it pops out a random element
# clear() : it empties the set
print(st1)
# st1.remove(("kiwi", "strawberry"))
st1.discard(("Kiwi", "strawberry"))
print(st1)
```

```python
# loops in sets
for val in st1:
    print(val)
```

### join sets

```python
# union() and update() and | : join all the items in selected sets
# intersection() : the only items present in both set 
# difference() : the items which are in first set that are not in second set
# symmetric_difference() : it keeps all the items except duplicates

A1 = {"apple", "banana", "cherry", "Dragon_fruit"}
A2 = {"Kiwi", "strawberry", "lichies", "mango"}
print("A1 \t: ", A1)
print("A2 \t: ", A2)

# union() and update() : 
#    we can use set1.union(set2) or set1.union(set2, set3, set4) or set1 = set1|set2
A3 = A1|A2
print("A3 \t: ", A3)
A4 = A1.union(A3)
print("A4 \t: ", A4)


# joining a set and tuple using union()
t1 = ("fruit1", "fruit2", "fruit3")
TS = A1.union(t1)
print("TS \t: ", TS)


# intersection () and & :
#     return the elements which are common in both sets
print("A1 \t: ",A1)
print("A3 \t: ",A3)
Ainter = A1.intersection(A3)
print("Ainter \t: ", Ainter)

A_inter_sym = A2&A4
print("A_inter_sym : ", A_inter_sym)


# difference() and - (-ve) :
#     the items which are in the first set that are not in the second set
D1 = {"123", "423", 342, 893, 990.2}
D2 = {"423", 342, "Chandu"}
A_diff = D1 - D2
print("A_diff \t: ", A_diff)

A_diff2 = D2.difference(D1)
print("A_diff2 : ", A_diff2)

```

# Dictionaries : {"Key" : "Value"}

```python
# dictionaries store the data in key: value pairs 
# dictionaries are mutable
# the key should be unique but the values can be duplicate and also the values can have different datatypes


dict_c1 = {
    "brand": "Ford",
    "model": "Mustang",
    "start_year": 1964,
    "end_year": 2020,
    "places": ["place1", "place2", "place3"]
}
print(dict_c1)

# len() : length of the dictionary
print(f"length of dict_c1 :  {len(dict_c1)}")

# dictionary constructor:
dict_c2 = dict(first_name = "Chandu", last_name = "M", Age = "23", Mail = "c#m@gmial.com")
print(dict_c2)
```

### Access dictionary items

```python
#  we can access the elements by using keys as index or using .get() method
# if we use ["key"] : if the key is not present in the dict it raise an error
# if we use .get() : if the key is not present in the dict it return "none"

value1 = dict_c1["brand"]
print(value1)
value2 = dict_c1.get("end_year")
print(value2)


# keys() : to get all the keys 
print(f"{dict_c1.keys()}")
# values() : to get all the values 
print(f"\n{dict_c1.values()}")

# items() : to get all the items 
print(f"\n{dict_c1.items()}")
```

### changing the values

```python
#  using keys
print(dict_c2)
dict_c2["last_name"] = "MJ"
print(dict_c2)

dict_c2.update({"Mail":"cm1j@gmail.com"})
print(dict_c2)
```

### adding items  

```python
print(dict_c2)
dict_c2["Mobile"] = 123456
print(dict_c2)
```

```python
print(dict_c1)
dict_c1.update({"color":"Black"})
print(dict_c1)
```

### remove items

```python
# pop() : it removes the specified item
dict_c2.pop("last_name")

# popitem() : it removes the random item from the dict
dict_c2.popitem()

# del 
del dict_c2["Age"]

# clear() : 
#     it clears the entire dictionary

print(dict_c2)
```

### Looping

```python
#  for in
for i in dict_c1:
    print(i)
    
print()

# only keys
for j in dict_c1.keys():
    print(j)
    
print()

# only values
for k in dict_c1:
    print(dict_c1[k])
print()

for k in dict_c1.values():
    print(k)
print()

# items
for l in dict_c1.items():
    print(l)
print()

for i, j in dict_c1.items():
    print(i, " = ", j)
```

### Copy of dictionary

```python
#  we cannot copy the dictinory as dict_2 = dict_1 we need to use inbuilt copy() method
new_dict_1= dict_c1.copy()
print("new_dict_1 :\n",new_dict_1)

new_dict_2 = dict(dict_c1)
print("new_dict_1 :\n",new_dict_2)
```

### Nested dictionaries

```python
Sec_A = {
    "std1":{
        "std_name" : "Chandu",
        "ID" : 35,
        "Mail" : "chandu@mmail.com",
        "subjects": ["maths", "physics", "chemistry"],
        "grade" : "A"
    },
    "std2":{
        "std_name" : "krishna",
        "ID" : 23,
        "Mail" : "krishna@mmail.com",
        "subjects": ["maths", "physics", "chemistry"],
        "grade" : "A"
    },
    "std3":{
        "std_name" : "Sai",
        "ID" : 6,
        "Mail" : "sai@mmail.com",
        "subjects": ["maths", "physics", "chemistry"],
        "grade" : "A"
    },
    "std4":{
        "std_name" : "abhijeeth",
        "ID" : 57,
        "Mail" : "abhi@mmail.com",
        "subjects": ["maths", "physics", "chemistry"],
        "grade" : "A"
    },
    "std5":{},
    "std6":{}
}


Sec_A
```

```python
#  to access items from nested dict we need to use [][]  it is same as nested list
print(Sec_A["std1"]["Mail"])

print()
# loop through nested dict
for k, v in Sec_A.items():
    print(k, " - ")
    
    for i in v:
        print(i, " : ", v[i] )
    print()
```
