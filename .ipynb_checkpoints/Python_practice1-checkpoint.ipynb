```python
# Mark Even and Odd - Python
"""
Given a positive integer x. Your task is to check, if it is even or odd (Any number that gives zero as remainder when divided by 2 is an even number)

Examples

Input: x = 4
Output: Even
Input: x = 5
Output: Odd

"""

def check_num(inp : int)->bool:
    if inp%2 == 0:
        return True
    else:
        return False
inpt  = int(input("Enter a random number"))
print(f"{'Even' if check_num(inpt) else 'Odd'}")
```

```python
# Check the status
"""
Given two integer variables a and b, and a boolean variable flag. The task is to check the status and return accordingly.

Return True for the following cases:
Either a or b (not both) is non-negative and the flag is false.
Both a and b are negative and the flag is true.

Otherwise, return false.

Example: 

Input:
a = 1
b = -1
flag = False
Output:
True
"""


def check_status(a, b, flag):
    ##Your code here
    if (a<0<b or b<0<a) and flag == False:
        return True
    elif a<0 and b<0 and flag == True:
        return True
    else:
        return False
    
print(f"{'True' if (check_status(2, 1, True)) else 'False'}")
```

```python
# Trouble with Angry Friends

"""
There are two friends, John and Smith, 
and the parameters j_angry and s_angry to indicate if each is angry. 
You are in trouble if both of them are angry or no one of them is angry.

returns true if you are in trouble, else return false

Example 1:

Input:
j_angry = True, s_angry = True
Output:
True
"""

def trouble(j_angry, s_angry):
    return j_angry == s_angry

print(f"{'Trouble' if trouble(True, False) else 'Safe'}")
```

```python
# Check for Equal Occurrences of ‘cat’ and ‘hat’ in a String
"""
You are given a string str, you need to return True if  the words "cat" and "hat" appear same number of times in str, 
otherwise return False.
Note: str contains only lowercase English alphabets.

Example 1:
Input:
str = catinahat
Output:
True
Explanation:
cat and hat both are present
1 number of times.
"""


def occ_cat_hat(str):
    return str.count('cat') == str.count('hat')

print(f"{'Same' if occ_cat_hat('catinahatwhichisinanotherhat') else 'Different'}")
```

```python
# Conditional Boolean Evaluation
"""
You are given a number N, you need to print its multiplication table.
Note: Please go through the range function to understand why it's useful in for loops.

Example 1:
Input:
N = 5
Output:
5 10 15 20 25 30 35 40 45 50
"""

def print_table(num):
    for i in range(1, 11):
        print(i*num, end=" ")
print_table(4)
```

```python
# 
"""
You are given a string str, you need to print its characters at even indices(index starts at 0).
Note: Please go through the range function to understand how to jump 2 steps.

Example 1:
Input:
str = DoctorPhenomenal
Output:
DcoPeoea
"""

def even_ind_char(str):
    for i in range(0, len(str)-1, 2):
        print(str[i], end="-")
even_ind_char("DoctorPhenomenal")
```

```python
# Print in dec order using while
"""
Given a number x, the task is to print the numbers from x to 0 in decreasing order in a single line.
Example:
Input:
x = 3
Output:
3 2 1 0
"""

def decrease_order(num):
    while(num>=0):
        print(num, end=" ")
        num-=1
        
decrease_order(10)
```

```python
# Jumping through While

"""
Given a positive integer x, the task is to print the numbers from 1 to x 
in the order as 1^2, 2^2, 3^2, 4^2, 5^2, ... (in increasing order).
Example:

Input:
x = 10
Output:
1 4 9
Explanation:
From 1 to 10, numbers in powers
of 2 are, 1^2, 2^2, 3^2 as 1, 4 and 9.

"""

def squares(num):
    i =1
    while(i<=num):
        val =  i*i
        if val>num:
            break
        print (val, end = " ")
        i+=1
        
        
        
squares(100)
```

```python

"""
You are given a number n. The number n can be negative or positive. 
If n is negative, print numbers from n to 0 by adding 1 to n in the neg function. 
If positive, print numbers from n-1 to 0 by subtracting 1 from n in the pos function.

Example 1:
Input:
n = -3
Output:
-3 -2 -1 0

Example 2:
Input:
n = 4
Output:
3 2 1 0
"""

def print_num(num):
    if num<0:
        while(num<=0):
            print(num, end=" ")
            num+=1
    elif num>0:
        num-=1
        while(num>=0):
            print(num, end=" ")
            num-=1
    else:
        print(num)
        
print_num(6)
print()
print_num(-6)
print()
print_num(0)
```
